```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate
from tensorflow.keras.optimizers import Adam

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # Forcer usage CPU
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        img = Image.open(img_path).convert("RGB").resize(image_size)
        mask = Image.open(mask_path).convert("L").resize(image_size, resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    return images, masks


def build_light_unet(input_shape=(256, 256, 3), num_classes=8):
    inputs = Input(shape=input_shape)

    c1 = Conv2D(32, 3, activation='relu', padding='same')(inputs)
    p1 = MaxPooling2D()(c1)
    c2 = Conv2D(64, 3, activation='relu', padding='same')(p1)
    p2 = MaxPooling2D()(c2)

    b = Conv2D(128, 3, activation='relu', padding='same')(p2)

    u1 = UpSampling2D()(b)
    m1 = concatenate([u1, c2])
    c3 = Conv2D(64, 3, activation='relu', padding='same')(m1)
    u2 = UpSampling2D()(c3)
    m2 = concatenate([u2, c1])
    c4 = Conv2D(32, 3, activation='relu', padding='same')(m2)

    outputs = Conv2D(num_classes, 1, activation='softmax')(c4)

    return Model(inputs, outputs)


def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    num_classes = 8
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


def train_unet_cityscapes(
    image_size=(256, 256),
    batch_size=4,
    epochs=10,
    data_dir="/home/romain/work/projet8/notebooks/data",
    experiment_name="unet-segmentation-8classes",
    model_name="unet_light_cityscapes"
):
    import mlflow.tensorflow
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🧠 Création du modèle U-Net léger...")
        model = build_light_unet(input_shape=(*image_size, 3), num_classes=8)
        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Démarrage de l'entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle localement au format `.keras`...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        print("📦 Génération d’un input_example et de la signature...")
        # input_example pour inférer la signature
        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement du modèle dans MLflow (onglet 'Models')...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            
            signature=signature
        )

    return model, history

```

```python
model, history = train_unet_cityscapes(
    image_size=(256, 256),
    batch_size=4,
    epochs=12,
    data_dir="/srv/picture_segmentation/projet8_openclassrooms/notebooks/data",
    experiment_name="segmentation_cityscapes",
    model_name="unet_cityscapes_cpu"
)
```

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate, BatchNormalization
from tensorflow.keras.optimizers import Adam
from concurrent.futures import ThreadPoolExecutor
from functools import partial

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # Forcer usage CPU
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    return images, masks


# === UNet modulaire ===
def double_conv(x, filters):
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    return x

def build_unet(input_shape=(256, 256, 3), num_classes=8, depth=2, base_filters=32):
    inputs = Input(shape=input_shape)

    # Encodeur
    encoders = []
    x = inputs
    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = MaxPooling2D()(x)

    # Bottleneck
    x = double_conv(x, base_filters * (2 ** depth))

    # Decodeur
    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = UpSampling2D()(x)
        x = concatenate([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = Conv2D(num_classes, 1, activation='softmax')(x)
    return Model(inputs, outputs)


# === Fonctions de loss et metrics ===

def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    num_classes = 8
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


# === Entraînement ===

def train_unet_cityscapes(
    image_size=(512, 512),         # 👈 taille image ajustée
    batch_size=4,
    epochs=10,
    data_dir="/home/romain/work/projet8/notebooks/data",
    experiment_name="unet-segmentation-8classes",
    model_name="unet_light_cityscapes",
    depth=3,                       # 👈 profondeur configurable
    base_filters=64               # 👈 capacité du modèle
):
    import mlflow.tensorflow
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8_true/train"),
        image_size=image_size
    )

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8_true/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🧠 Création du modèle U-Net paramétrable...")
        model = build_unet(
            input_shape=(*image_size, 3),
            num_classes=8,
            depth=depth,
            base_filters=base_filters
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Démarrage de l'entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle localement au format `.keras`...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        print("📦 Génération d’un input_example et de la signature...")
        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement du modèle dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

# Inclusion pixel 0

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate, BatchNormalization
from tensorflow.keras.optimizers import Adam
from concurrent.futures import ThreadPoolExecutor
from functools import partial

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)


def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    return images, masks


def double_conv(x, filters):
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    return x


def build_unet(input_shape=(256, 256, 3), num_classes=8, depth=2, base_filters=32):
    inputs = Input(shape=input_shape)

    encoders = []
    x = inputs
    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = MaxPooling2D()(x)

    x = double_conv(x, base_filters * (2 ** depth))

    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = UpSampling2D()(x)
        x = concatenate([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = Conv2D(num_classes, 1, activation='softmax')(x)
    return Model(inputs, outputs)


# === Loss et metrics sans masquage ===

def mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    num_classes = 8
    iou_list = []

    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


# === Entraînement ===

def train_unet_cityscapes(
    image_size=(512, 512),
    batch_size=4,
    epochs=10,
    data_dir="/home/romain/work/projet8/notebooks/data",
    experiment_name="unet-segmentation-8classes",
    model_name="unet_light_cityscapes",
    depth=3,
    base_filters=64
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8_true/train"),
        image_size=image_size
    )

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8_true/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🧠 Création du modèle U-Net paramétrable...")
        model = build_unet(
            input_shape=(*image_size, 3),
            num_classes=8,
            depth=depth,
            base_filters=base_filters
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=tf.keras.losses.SparseCategoricalCrossentropy(),
            metrics=["accuracy", mean_iou, dice_coef]
        )

        print("🚀 Démarrage de l'entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle localement au format `.keras`...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        print("📦 Génération d’un input_example et de la signature...")
        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement du modèle dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate, BatchNormalization
from tensorflow.keras.optimizers import Adam
from concurrent.futures import ThreadPoolExecutor
from functools import partial
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # Forcer usage CPU
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")





def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    return images, masks


# === UNet modulaire ===
def double_conv(x, filters):
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    return x

def build_unet(input_shape=(256, 256, 3), num_classes=8, depth=2, base_filters=32):
    inputs = Input(shape=input_shape)

    # Encodeur
    encoders = []
    x = inputs
    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = MaxPooling2D()(x)

    # Bottleneck
    x = double_conv(x, base_filters * (2 ** depth))

    # Decodeur
    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = UpSampling2D()(x)
        x = concatenate([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = Conv2D(num_classes, 1, activation='softmax')(x)
    return Model(inputs, outputs)


# === Fonctions de loss et metrics ===

def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    num_classes = 8
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


# === Entraînement ===

def train_unet_cityscapes(
    image_size=(512, 512),         # 👈 taille image ajustée
    batch_size=4,
    epochs=10,
    data_dir="/home/romain/work/projet8/notebooks/data",
    experiment_name="unet-segmentation-8classes",
    model_name="unet_light_cityscapes",
    depth=3,                       # 👈 profondeur configurable
    base_filters=64               # 👈 capacité du modèle
):
    import mlflow.tensorflow
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit_aug4/train"),
        os.path.join(data_dir, "masks_8classes_aug4/train"),
        image_size=image_size
    )

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🧠 Création du modèle U-Net paramétrable...")
        model = build_unet(
            input_shape=(*image_size, 3),
            num_classes=8,
            depth=depth,
            base_filters=base_filters
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Démarrage de l'entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle localement au format `.keras`...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        print("📦 Génération d’un input_example et de la signature...")
        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement du modèle dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import tensorflow_hub as hub
from tensorflow.keras.optimizers import Adam
import mlflow
import mlflow.tensorflow

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        img = Image.open(img_path).convert("RGB").resize(image_size)
        mask = Image.open(mask_path).convert("L").resize(image_size, resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    return images, masks


# === LOSS et METRIQUES avec upsampling ===

def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')

    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')

    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')

    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


# === Entraînement avec DeepLabV3 ===

def train_deeplabv3_cityscapes(
    image_size=(256, 256),
    batch_size=4,
    epochs=10,
    data_dir="/path/to/data",
    experiment_name="deeplabv3_cityscapes_exp",
    model_name="deeplabv3_cityscapes_model"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🔌 Chargement du modèle DeepLabV3 depuis TensorFlow Hub...")
        input_tensor = tf.keras.layers.Input(shape=(*image_size, 3), dtype=tf.float32)
        base_model = hub.KerasLayer(
            "https://tfhub.dev/tensorflow/deeplabv3/1",
            trainable=True
        )
        output = base_model(input_tensor)
        output = output['default']  # Format dict
        model = tf.keras.Model(inputs=input_tensor, outputs=output)

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
model, history = train_deeplabv3_cityscapes(
    image_size=(256, 256),                         # Taille des images
    batch_size=4,                                  # Taille du batch
    epochs=3,                                     # Nombre d'epochs
    data_dir="/srv/picture_segmentation/projet8_openclassrooms/notebooks/data",  # Dossier de données
    experiment_name="deeplabv3_cityscapes_exp",    # Nom de l'expérience MLflow
    model_name="deeplabv3_cityscapes_model"        # Nom du modèle à enregistrer
)
```

# MobileNetV2

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.layers import Conv2D, UpSampling2D, Input
from tensorflow.keras.models import Model
import mlflow
import mlflow.tensorflow

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        # Note l'inversion (largeur, hauteur) pour PIL resize
        img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
        mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    print(f"🖼️ Forme des images chargées : {images.shape}")

    return images, masks



def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    y_pred = tf.image.resize(y_pred, tf.shape(y_true)[1:3], method='bilinear')
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


def build_deeplabv3plus_mobilenetv2(image_size=(256, 256), num_classes=8, weights_path=None):
    # Input
    inputs = Input(shape=(*image_size, 3))

    # Backbone MobileNetV2
    backbone = MobileNetV2(input_tensor=inputs, include_top=False, weights=None)  # pas de poids auto
    if weights_path:
        backbone.load_weights(weights_path)

    # Features extraction
    x = backbone.output  # shape réduit (1/32)

    # Simple DeepLabV3+ head (à adapter plus finement si besoin)
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = UpSampling2D(size=(4,4), interpolation='bilinear')(x)  # remonter à 1/8
    x = Conv2D(num_classes, 1, padding='same')(x)
    x = UpSampling2D(size=(8,8), interpolation='bilinear')(x)  # remonter à taille originale

    model = Model(inputs=inputs, outputs=x)
    return model


def train_deeplabv3plus_cityscapes(
    image_size=(256, 256),
    batch_size=2,
    epochs=5,
    data_dir="/path/to/data",
    experiment_name="deeplabv3plus_cityscapes_exp",
    model_name="deeplabv3plus_cityscapes_model",
    mobilenet_weights_path=None
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🔌 Construction du modèle DeepLabV3+ avec MobileNetV2 backbone local...")
        model = build_deeplabv3plus_mobilenetv2(
            image_size=image_size,
            num_classes=8,
            weights_path=mobilenet_weights_path
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history


# Exemple d'appel (à adapter)
# model, history = train_deeplabv3plus_cityscapes(
#     image_size=(256, 256),
#     batch_size=2,
#     epochs=5,
#     data_dir="/chemin/vers/tes/données",
#     experiment_name="exp_local_mobilenetv2",
#     model_name="deeplabv3plus_local_mobilenetv2",
#     mobilenet_weights_path="/chemin/vers/mobilenet_v2_weights_tf_dim_ordering_tf_kernels_1.0_256_no_top.h5"
# )
```

# DeepLabV3

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
import mlflow
import mlflow.tensorflow
from tensorflow.keras.layers import Input, Conv2D, UpSampling2D, Concatenate
from tensorflow.keras.models import Model
from tensorflow.keras.applications import ResNet50


# --- Configuration environnement ---
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # CPU only (change si GPU dispo)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")

# --- Chargement images + masques ---
def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        # Note l'inversion (largeur, hauteur) pour PIL resize
        img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
        mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    print(f"🖼️ Forme des images chargées : {images.shape}")

    return images, masks


# --- Fonctions loss et metrics personnalisés ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)

def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)

def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)

# --- DeepLabV3+ avec ResNet50 backbone (simplifié) ---
from tensorflow.keras.layers import Input
from tensorflow.keras.models import Model
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers

def deeplabv3plus_resnet50(image_size=(256, 256), num_classes=8, weights_path=None, backbone_weights_path=None):
    inputs = Input(shape=(*image_size, 3))

    # Charger ResNet50 sans les poids du top, et sans poids par défaut (None)
    base_model = ResNet50(input_tensor=inputs, include_top=False, weights=None)
    if backbone_weights_path:
        base_model.load_weights(backbone_weights_path)

    # Extraction d'une couche basse résolution pour fusion (low level feature)
    low_level_feat = base_model.get_layer('conv2_block3_out').output  # (None, 64, 64, 512)
    low_level_feat = layers.Conv2D(48, 1, padding='same', activation='relu')(low_level_feat)  # (None, 64, 64, 48)

    # Traitement du output final de ResNet50
    x = base_model.output  # (None, 8, 8, 2048)
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)  # (None, 8, 8, 256)

    # Upsample x de 8x pour matcher low_level_feat (64x64)
    x = layers.UpSampling2D(size=(8, 8), interpolation='bilinear')(x)  # (None, 64, 64, 256)

    # Concaténation des features
    x = layers.Concatenate()([x, low_level_feat])  # (None, 64, 64, 256 + 48 = 304)

    # Bloc convolutif après concaténation
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)

    # Upsample final 4x pour passer à la résolution finale (256x256)
    x = layers.UpSampling2D(size=(4, 4), interpolation='bilinear')(x)  # (None, 256, 256, 256)

    # Couche de sortie avec softmax sur les classes
    outputs = layers.Conv2D(num_classes, 1, padding='same', activation='softmax')(x)  # (None, 256, 256, num_classes)

    model = Model(inputs=inputs, outputs=outputs)

    # Charger les poids du modèle complet si fournis
    if weights_path:
        model.load_weights(weights_path)

    return model





# --- Fonction d'entraînement complète avec MLflow ---
def train_deeplabv3plus_resnet50_local(
    image_size=(256,256),
    batch_size=2,
    epochs=5,
    data_dir="/path/to/data",
    model_weights_path=None,
    backbone_weights_path=None,
    model_name="deeplabv3plus_cityscapes_model",
    experiment_name="deeplabv3plus_resnet50_local_exp"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🔌 Construction du modèle DeepLabV3+ ResNet50 local...")
        model = deeplabv3plus_resnet50(
            image_size=image_size,
            num_classes=8,
            weights_path=model_weights_path
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle...")
        keras_model_path = "deeplabv3plus_resnet50_local.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name="deeplabv3plus_resnet50_local",
            input_example=input_example,
            signature=signature
        )

    return model, history

# --- Exemple d'appel --- à adapter ---
# model, history = train_deeplabv3plus_resnet50_local(
#     image_size=(256,256),
#     batch_size=2,
#     epochs=5,
#     data_dir="/chemin/vers/donnees",
#     model_weights_path="/chemin/vers/model.weights.h5",
#     experiment_name="exp_deeplabv3plus_resnet50_local"
# )
```

# Resnet 101 + DeeplabV3

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
import mlflow
import mlflow.tensorflow
from tensorflow.keras.layers import Input, Conv2D, UpSampling2D, Concatenate
from tensorflow.keras.models import Model



# --- Configuration environnement ---
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # CPU only (change si GPU dispo)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")

# --- Chargement images + masques ---
def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        # Note l'inversion (largeur, hauteur) pour PIL resize
        img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
        mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    print(f"🖼️ Forme des images chargées : {images.shape}")

    return images, masks


# --- Fonctions loss et metrics personnalisés ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)

def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)

def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)

# --- DeepLabV3+ avec ResNet50 backbone (simplifié) ---
from tensorflow.keras.layers import Input
from tensorflow.keras.models import Model
from tensorflow.keras import layers

from tensorflow.keras.applications import ResNet101

def deeplabv3plus_resnet101(image_size=(256, 256), num_classes=8, weights_path=None, backbone_weights_path=None):
    inputs = Input(shape=(*image_size, 3))

    base_model = ResNet101(input_tensor=inputs, include_top=False, weights=None)
    if backbone_weights_path:
        base_model.load_weights(backbone_weights_path)

    low_level_feat = base_model.get_layer('conv2_block3_out').output
    low_level_feat = layers.Conv2D(48, 1, padding='same', activation='relu')(low_level_feat)

    x = base_model.output
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.UpSampling2D(size=(8, 8), interpolation='bilinear')(x)

    x = layers.Concatenate()([x, low_level_feat])
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)

    x = layers.UpSampling2D(size=(4, 4), interpolation='bilinear')(x)
    outputs = layers.Conv2D(num_classes, 1, padding='same', activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)

    if weights_path:
        model.load_weights(weights_path)

    return model




# --- Fonction d'entraînement complète avec MLflow ---
def train_deeplabv3plus_resnet101_local(
    image_size=(256,256),
    batch_size=2,
    epochs=5,
    data_dir="/path/to/data",
    model_weights_path=None,
    backbone_weights_path=None,
    model_name="deeplabv3plus_cityscapes_model_resnet101",
    experiment_name="deeplabv3plus_resnet101_local_exp"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🔌 Construction du modèle DeepLabV3+ ResNet101 local...")
        model = deeplabv3plus_resnet101(
            image_size=image_size,
            num_classes=8,
            weights_path=model_weights_path,
            backbone_weights_path=backbone_weights_path
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle...")
        keras_model_path = "deeplabv3plus_resnet101_local.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name="deeplabv3plus_resnet101_local",
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
import mlflow
import mlflow.tensorflow
from tensorflow.keras.layers import Input, Conv2D, UpSampling2D, Concatenate
from tensorflow.keras.models import Model

# --- Configuration environnement ---
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # CPU only (change si GPU dispo)
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")

# --- Chargement images + masques ---
def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256)):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques filtrés trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    images, masks = [], []
    for img_path, mask_path in zip(image_paths, mask_paths):
        # Note l'inversion (largeur, hauteur) pour PIL resize
        img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
        mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)

        images.append(np.array(img) / 255.0)
        masks.append(np.array(mask))

    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    print(f"🖼️ Forme des images chargées : {images.shape}")

    return images, masks


# --- Data Augmentation ---
def augment_image(img, mask):
    if tf.random.uniform(()) > 0.5:
        img = tf.image.flip_left_right(img)
        mask = tf.image.flip_left_right(mask)
    if tf.random.uniform(()) > 0.5:
        img = tf.image.random_brightness(img, max_delta=0.1)
    return img, mask

def dataset_from_numpy(images, masks, batch_size, augment=False, image_size=(256, 256)):
    def preprocess(img, mask):
        img = tf.cast(img, tf.float32)
        mask = tf.cast(mask, tf.uint8)
        if augment:
            img, mask = augment_image(img, mask)
        img.set_shape([image_size[0], image_size[1], 3])
        mask.set_shape([image_size[0], image_size[1], 1])
        return img, mask

    ds = tf.data.Dataset.from_tensor_slices((images, masks))
    ds = ds.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    ds = ds.shuffle(100).batch(batch_size).prefetch(tf.data.AUTOTUNE)
    return ds

# --- Fonctions loss et metrics personnalisés ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)

def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    for i in range(8):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)

def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    for i in range(8):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)

# --- DeepLabV3+ avec ResNet101 ---
def deeplabv3plus_resnet101(image_size=(256, 256), num_classes=8, weights_path=None, backbone_weights_path=None):
    inputs = Input(shape=(*image_size, 3))
    base_model = tf.keras.applications.ResNet101(input_tensor=inputs, include_top=False, weights=None)
    if backbone_weights_path:
        base_model.load_weights(backbone_weights_path)
    low_level_feat = base_model.get_layer('conv2_block3_out').output
    low_level_feat = layers.Conv2D(48, 1, padding='same', activation='relu')(low_level_feat)
    x = base_model.output
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.UpSampling2D(size=(8, 8), interpolation='bilinear')(x)
    x = layers.Concatenate()([x, low_level_feat])
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.Conv2D(256, 3, padding='same', activation='relu')(x)
    x = layers.UpSampling2D(size=(4, 4), interpolation='bilinear')(x)
    outputs = layers.Conv2D(num_classes, 1, padding='same', activation='softmax')(x)
    return Model(inputs=inputs, outputs=outputs)

# --- Entraînement complet ---
def train_deeplabv3plus_resnet101_local(
    image_size=(256,256),
    batch_size=2,
    epochs=5,
    data_dir="/path/to/data",
    model_weights_path=None,
    backbone_weights_path=None,
    model_name="deeplabv3plus_resnet101_model",
    experiment_name="deeplabv3plus_resnet101_exp"
):
    from mlflow.models.signature import infer_signature
    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/train"),
        os.path.join(data_dir, "masks_8classes/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    train_ds = dataset_from_numpy(train_images, train_masks, batch_size=batch_size, augment=True, image_size=image_size)
    val_ds = dataset_from_numpy(val_images, val_masks, batch_size=batch_size, augment=False, image_size=image_size)

    with mlflow.start_run():
        model = deeplabv3plus_resnet101(image_size=image_size, weights_path=model_weights_path, backbone_weights_path=backbone_weights_path)
        model.compile(optimizer=Adam(1e-3),
                      loss=sparse_categorical_crossentropy_ignore_255,
                      metrics=["accuracy", masked_mean_iou, masked_dice_coef])

        history = model.fit(train_ds, validation_data=val_ds, epochs=epochs)

        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        mlflow.tensorflow.log_model(model=model,
                                    artifact_path="model",
                                    registered_model_name=model_name,
                                    input_example=input_example,
                                    signature=signature)
    return model, history
```

```python
model, history = train_deeplabv3plus_resnet101_local(
    image_size=(256, 256),  # Ou (512, 512), etc.
    batch_size=4,
    epochs=4,
    data_dir="/srv/picture_segmentation/projet8_openclassrooms/notebooks/data",  # Remplace par ton chemin local
    model_weights_path=None,              # Ou chemin vers des poids préentraînés `.h5`
    backbone_weights_path="/srv/picture_segmentation/projet8_openclassrooms/notebooks/resnet101_weights_tf_dim_ordering_tf_kernels_notop.weights.h5",           # Idem, ou None si pas de prétraining
    model_name="deeplabv3plus_resnet101_v1",
    experiment_name="exp_cityscapes_resnet101_256x256_data_augmentation"
)
```

# Data augmentation sur le disque

```python
import os
import numpy as np
from PIL import Image, ImageEnhance
import random
from glob import glob

def augment_image_and_mask(image, mask):
    # Rotation
    angle = random.choice([0, 90, 180, 270])
    image = image.rotate(angle)
    mask = mask.rotate(angle)

    # Flip horizontal / vertical
    if random.random() > 0.5:
        image = image.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)
    if random.random() > 0.5:
        image = image.transpose(Image.FLIP_TOP_BOTTOM)
        mask = mask.transpose(Image.FLIP_TOP_BOTTOM)

    # Brightness / Contrast / Color
    enhancers = [
        ImageEnhance.Brightness,
        ImageEnhance.Contrast,
        ImageEnhance.Color,
    ]
    for enhancer in enhancers:
        factor = random.uniform(0.8, 1.2)
        image = enhancer(image).enhance(factor)

    # Zoom crop (central or random)
    if random.random() > 0.5:
        w, h = image.size
        scale = random.uniform(0.9, 1.0)
        new_w, new_h = int(w * scale), int(h * scale)
        left = random.randint(0, w - new_w)
        top = random.randint(0, h - new_h)
        image = image.crop((left, top, left + new_w, top + new_h)).resize((w, h), Image.BILINEAR)
        mask = mask.crop((left, top, left + new_w, top + new_h)).resize((w, h), Image.NEAREST)

    # Bruit (uniquement image)
    if random.random() > 0.5:
        image_np = np.array(image)
        noise = np.random.normal(0, 5, image_np.shape).astype(np.uint8)
        image_np = np.clip(image_np + noise, 0, 255).astype(np.uint8)
        image = Image.fromarray(image_np)

    return image, mask

def augment_and_save_images_on_disk(
    image_dir,
    mask_dir,
    output_image_dir,
    output_mask_dir,
    augmentations_per_image=9
):
    os.makedirs(output_image_dir, exist_ok=True)
    os.makedirs(output_mask_dir, exist_ok=True)

    image_paths = sorted(glob(os.path.join(image_dir, "*", "*.png")))
    mask_paths = sorted(glob(os.path.join(mask_dir, "*", "*_gtFine_8class.png")))

    print(f"🔎 {len(image_paths)} images et {len(mask_paths)} masques à augmenter...")

    for img_path, mask_path in zip(image_paths, mask_paths):
        city = os.path.basename(os.path.dirname(img_path))
        os.makedirs(os.path.join(output_image_dir, city), exist_ok=True)
        os.makedirs(os.path.join(output_mask_dir, city), exist_ok=True)

        base_filename = os.path.splitext(os.path.basename(img_path))[0]
        img = Image.open(img_path).convert("RGB")
        mask = Image.open(mask_path).convert("L")

        # Sauvegarde originale (optionnel)
        # img.save(os.path.join(output_image_dir, city, f"{base_filename}_orig.png"))
        # mask.save(os.path.join(output_mask_dir, city, f"{base_filename}_orig_gtFine_8class.png"))

        for i in range(augmentations_per_image):
            img_aug, mask_aug = augment_image_and_mask(img.copy(), mask.copy())
            suffix = f"aug{i+1:02d}"
            new_img_name = f"{base_filename}_{suffix}.png"
            new_mask_name = f"{base_filename}_{suffix}_gtFine_8class.png"

            img_aug.save(os.path.join(output_image_dir, city, new_img_name))
            mask_aug.save(os.path.join(output_mask_dir, city, new_mask_name))

    print(f"✅ Augmentation terminée : {augmentations_per_image}x par image.")
```

```python
import os
import random
from PIL import Image, ImageEnhance, ImageOps, ImageFilter
import numpy as np
from glob import glob
from concurrent.futures import ProcessPoolExecutor

def add_random_noise(image):
    np_img = np.array(image)
    noise = np.random.normal(0, 5, np_img.shape).astype(np.int16)
    noisy_img = np.clip(np_img + noise, 0, 255).astype(np.uint8)
    return Image.fromarray(noisy_img)

def apply_random_transformations(img, mask):

    if random.random() < 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)

    if random.random() < 0.5:
        img = img.transpose(Image.FLIP_TOP_BOTTOM)
        mask = mask.transpose(Image.FLIP_TOP_BOTTOM)

    if random.random() < 0.5:
        angle = random.choice([90, 180, 270])
        img = img.rotate(angle, resample=Image.BILINEAR)
        mask = mask.rotate(angle, resample=Image.NEAREST)

    if random.random() < 0.5:
        scale = random.uniform(0.9, 1.1)
        w, h = img.size
        new_w, new_h = int(w * scale), int(h * scale)

        img = img.resize((new_w, new_h), resample=Image.BILINEAR)
        mask = mask.resize((new_w, new_h), resample=Image.NEAREST)

        if scale < 1.0:
            pad_w = (w - new_w) // 2
            pad_h = (h - new_h) // 2
            img = ImageOps.expand(img, border=(pad_w, pad_h, w - new_w - pad_w, h - new_h - pad_h), fill=0)
            mask = ImageOps.expand(mask, border=(pad_w, pad_h, w - new_w - pad_w, h - new_h - pad_h), fill=255)
        else:
            left = (new_w - w) // 2
            top = (new_h - h) // 2
            img = img.crop((left, top, left + w, top + h))
            mask = mask.crop((left, top, left + w, top + h))

    if random.random() < 0.3:
        factor = random.uniform(0.7, 1.3)
        img = ImageEnhance.Brightness(img).enhance(factor)

    if random.random() < 0.3:
        factor = random.uniform(0.7, 1.5)
        img = ImageEnhance.Contrast(img).enhance(factor)

    if random.random() < 0.2:
        radius = random.uniform(0.5, 1.5)
        img = img.filter(ImageFilter.GaussianBlur(radius=radius))

    if random.random() < 0.2:
        img = add_random_noise(img)

    return img, mask

def augment_one_pair(img_path, mask_path, out_img_dir, out_mask_dir, rel_path, n_aug=5):
    base_name = os.path.splitext(os.path.basename(img_path))[0]

    img = Image.open(img_path).convert("RGB")
    mask = Image.open(mask_path).convert("L")

    for i in range(n_aug):
        aug_img, aug_mask = apply_random_transformations(img, mask)

        # Créer les répertoires de sortie selon la structure relative
        sub_dir = os.path.dirname(rel_path)
        os.makedirs(os.path.join(out_img_dir, sub_dir), exist_ok=True)
        os.makedirs(os.path.join(out_mask_dir, sub_dir), exist_ok=True)

        out_img_path = os.path.join(out_img_dir, sub_dir, f"{base_name}_aug{i}.png")
        out_mask_path = os.path.join(out_mask_dir, sub_dir, f"{base_name}_aug{i}.png")

        aug_img.save(out_img_path)
        aug_mask.save(out_mask_path)


def augment_dataset_multicore(img_dir, mask_dir, out_img_dir, out_mask_dir, n_aug_per_image=5, num_workers=8):
    print(f"Début de l'augmentation avec {num_workers} workers...")
    os.makedirs(out_img_dir, exist_ok=True)
    os.makedirs(out_mask_dir, exist_ok=True)

    img_paths = sorted(glob(os.path.join(img_dir, "**/*.png"), recursive=True))
    mask_paths = sorted(glob(os.path.join(mask_dir, "**/*.png"), recursive=True))

    
    with ProcessPoolExecutor(max_workers=num_workers) as executor:
        futures = []
        for img_path, mask_path in zip(img_paths, mask_paths):
            rel_path = os.path.relpath(img_path, img_dir)
            futures.append(executor.submit(
                augment_one_pair,
                img_path, mask_path,
                out_img_dir, out_mask_dir,
                rel_path,
                n_aug_per_image
            ))

    print("Augmentation terminée.")
```

```python
import os
import random
from PIL import Image, ImageEnhance, ImageOps, ImageFilter
import numpy as np
from glob import glob
from concurrent.futures import ProcessPoolExecutor

def add_random_noise(image):
    np_img = np.array(image)
    noise = np.random.normal(0, 5, np_img.shape).astype(np.int16)
    noisy_img = np.clip(np_img + noise, 0, 255).astype(np.uint8)
    return Image.fromarray(noisy_img)

def apply_random_transformations(img, mask):
    if random.random() < 0.5:
        img = img.transpose(Image.FLIP_LEFT_RIGHT)
        mask = mask.transpose(Image.FLIP_LEFT_RIGHT)

    if random.random() < 0.5:
        img = img.transpose(Image.FLIP_TOP_BOTTOM)
        mask = mask.transpose(Image.FLIP_TOP_BOTTOM)

    if random.random() < 0.5:
        angle = random.choice([90, 180, 270])
        img = img.rotate(angle, resample=Image.BILINEAR)
        mask = mask.rotate(angle, resample=Image.NEAREST)

    if random.random() < 0.5:
        scale = random.uniform(0.9, 1.1)
        w, h = img.size
        new_w, new_h = int(w * scale), int(h * scale)

        img = img.resize((new_w, new_h), resample=Image.BILINEAR)
        mask = mask.resize((new_w, new_h), resample=Image.NEAREST)

        if scale < 1.0:
            pad_w = (w - new_w) // 2
            pad_h = (h - new_h) // 2
            img = ImageOps.expand(img, border=(pad_w, pad_h, w - new_w - pad_w, h - new_h - pad_h), fill=0)
            mask = ImageOps.expand(mask, border=(pad_w, pad_h, w - new_w - pad_w, h - new_h - pad_h), fill=255)
        else:
            left = (new_w - w) // 2
            top = (new_h - h) // 2
            img = img.crop((left, top, left + w, top + h))
            mask = mask.crop((left, top, left + w, top + h))

    if random.random() < 0.3:
        factor = random.uniform(0.7, 1.3)
        img = ImageEnhance.Brightness(img).enhance(factor)

    if random.random() < 0.3:
        factor = random.uniform(0.7, 1.5)
        img = ImageEnhance.Contrast(img).enhance(factor)

    if random.random() < 0.2:
        radius = random.uniform(0.5, 1.5)
        img = img.filter(ImageFilter.GaussianBlur(radius=radius))

    if random.random() < 0.2:
        img = add_random_noise(img)

    return img, mask

def augment_one_pair(img_path, mask_path, out_img_dir, out_mask_dir, rel_path, n_aug=5):
    base_name = os.path.splitext(os.path.basename(img_path))[0]
    sub_dir = os.path.dirname(rel_path)
    os.makedirs(os.path.join(out_img_dir, sub_dir), exist_ok=True)
    os.makedirs(os.path.join(out_mask_dir, sub_dir), exist_ok=True)

    img = Image.open(img_path).convert("RGB")
    mask = Image.open(mask_path).convert("L")

    # ➕ Sauvegarde de l'image et du masque originaux
    orig_img_path = os.path.join(out_img_dir, sub_dir, f"{base_name}_orig.png")
    orig_mask_path = os.path.join(out_mask_dir, sub_dir, f"{base_name}_orig.png")
    img.save(orig_img_path)
    mask.save(orig_mask_path)

    # 🔁 Génération des versions augmentées
    for i in range(n_aug):
        aug_img, aug_mask = apply_random_transformations(img, mask)
        out_img_path = os.path.join(out_img_dir, sub_dir, f"{base_name}_aug{i}.png")
        out_mask_path = os.path.join(out_mask_dir, sub_dir, f"{base_name}_aug{i}.png")
        aug_img.save(out_img_path)
        aug_mask.save(out_mask_path)

def augment_dataset_multicore(img_dir, mask_dir, out_img_dir, out_mask_dir, n_aug_per_image=5, num_workers=8):
    print(f"🔁 Début de l'augmentation avec {num_workers} workers...")
    os.makedirs(out_img_dir, exist_ok=True)
    os.makedirs(out_mask_dir, exist_ok=True)

    img_paths = sorted(glob(os.path.join(img_dir, "**/*.png"), recursive=True))
    mask_paths = sorted(glob(os.path.join(mask_dir, "**/*.png"), recursive=True))

    with ProcessPoolExecutor(max_workers=num_workers) as executor:
        futures = []
        for img_path, mask_path in zip(img_paths, mask_paths):
            rel_path = os.path.relpath(img_path, img_dir)
            futures.append(executor.submit(
                augment_one_pair,
                img_path, mask_path,
                out_img_dir, out_mask_dir,
                rel_path,
                n_aug_per_image
            ))

    print("✅ Augmentation terminée.")
```

### 📊 Tableau des transformations utilisées

| Transformation              | Nature                            | Effet sur les images                                  | Effet sur les masques                                  | Conserve les dimensions |
|----------------------------|-----------------------------------|-------------------------------------------------------|--------------------------------------------------------|-------------------------|
| `RandomFlip`               | Géométrique                        | Miroir horizontal                                    | Miroir horizontal                                      | ✅                      |
| `RandomRotation`           | Géométrique                        | Rotation jusqu’à ±15°                                | Rotation identique sur masque (interpolation NEAREST) | ✅                      |
| `RandomZoom`               | Géométrique                        | Zoom avant ou arrière                               | Zoom synchronisé (sans distorsion des classes)         | ✅                      |
| `RandomBrightness`         | Photométrique                      | Variation de la luminosité globale                   | Aucun effet (masque inchangé)                         | ✅                      |
| `RandomContrast`           | Photométrique                      | Variation du contraste global                        | Aucun effet                                            | ✅                      |
| `GaussianBlur`             | Filtrage / flou                    | Floute légèrement l’image                            | Aucun effet (ne s’applique pas au masque)             | ✅                      |
| `RandomCrop + Resize` (*)  | Géométrique                        | Découpe une région aléatoire, puis resize            | Idem, sur le masque                                    | ✅ (après resize)       |
| `ColorJitter`              | Photométrique (teinte, saturation) | Change couleur, teinte, saturation                   | Aucun effet                                            | ✅                      |

---

### 📌 Remarques

- Toutes les **transformations géométriques** (flip, rotation, zoom, crop) doivent être **appliquées à la fois à l’image et au masque**, avec **les mêmes paramètres**.
- Les **transformations photométriques** comme la luminosité ou le flou **ne s’appliquent qu’à l’image RGB**, **pas au masque**, qui doit rester strictement catégoriel.
- Les masques utilisent généralement une **interpolation NEAREST** pour préserver les labels (pas de flou ou de mélange de classes).
- Le **resize après crop** garantit que les images restent compatibles avec les tailles attendues dans le pipeline.


# Deeplab Resnet 101 data augmented dataset

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate, BatchNormalization
from tensorflow.keras.optimizers import Adam
from concurrent.futures import ThreadPoolExecutor
from functools import partial

os.environ["CUDA_VISIBLE_DEVICES"] = "-1"  # Forcer usage CPU
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    return images, masks


# === UNet modulaire ===
def double_conv(x, filters):
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    return x

def build_unet(input_shape=(256, 256, 3), num_classes=8, depth=2, base_filters=32):
    inputs = Input(shape=input_shape)

    # Encodeur
    encoders = []
    x = inputs
    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = MaxPooling2D()(x)

    # Bottleneck
    x = double_conv(x, base_filters * (2 ** depth))

    # Decodeur
    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = UpSampling2D()(x)
        x = concatenate([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = Conv2D(num_classes, 1, activation='softmax')(x)
    return Model(inputs, outputs)


# === Fonctions de loss et metrics ===

def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    num_classes = 8
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)


# === Entraînement ===

def train_unet_cityscapes(
    image_size=(512, 512),         # 👈 taille image ajustée
    batch_size=4,
    epochs=10,
    data_dir="/home/romain/work/projet8/notebooks/data",
    experiment_name="unet-segmentation-8classes",
    model_name="unet_light_cityscapes",
    depth=3,                       # 👈 profondeur configurable
    base_filters=64               # 👈 capacité du modèle
):
    import mlflow.tensorflow
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit_aug/train"),
        os.path.join(data_dir, "masks_8classes_aug/train"),
        image_size=image_size
    )

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8_true/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🧠 Création du modèle U-Net paramétrable...")
        model = build_unet(
            input_shape=(*image_size, 3),
            num_classes=8,
            depth=depth,
            base_filters=base_filters
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Démarrage de l'entraînement...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        print("💾 Sauvegarde du modèle localement au format `.keras`...")
        keras_model_path = f"{model_name}.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        print("📦 Génération d’un input_example et de la signature...")
        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement du modèle dans MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

# Sampling aléatoire d'un échantillon de taille arbitraire dans les données chargées

```python
import os
import random
import numpy as np
from PIL import Image
from glob import glob
from concurrent.futures import ThreadPoolExecutor
from functools import partial
import tensorflow as tf
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.layers import Input, Conv2D, UpSampling2D, Concatenate
from tensorflow.keras.models import Model
import mlflow
import mlflow.tensorflow

# --- Configuration environnement ---
os.environ["CUDA_VISIBLE_DEVICES"] = "-1"
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")

# --- Chargement images + masques ---
def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    if len(image_paths) == 0 or len(mask_paths) == 0:
        raise ValueError("❌ Aucun fichier trouvé. Vérifie les chemins et la structure des dossiers.")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    print(f"✅ Chargé {images.shape[0]} images et {masks.shape[0]} masques. Dimensions : {image_size}")
    return images, masks

# --- Dataset sampling ---
def create_sampled_dataset(images, masks, sample_size, batch_size, image_size=(256, 256)):
    def preprocess(img, mask):
        img = tf.cast(img, tf.float32)
        mask = tf.cast(mask, tf.uint8)
        img.set_shape([image_size[0], image_size[1], 3])
        mask.set_shape([image_size[0], image_size[1], 1])
        return img, mask

    def generator():
        idxs = random.sample(range(len(images)), min(sample_size, len(images)))
        for i in idxs:
            yield images[i], masks[i]

    dataset = tf.data.Dataset.from_generator(
        generator,
        output_signature=(
            tf.TensorSpec(shape=(image_size[0], image_size[1], 3), dtype=tf.float32),
            tf.TensorSpec(shape=(image_size[0], image_size[1], 1), dtype=tf.uint8),
        )
    )

    dataset = dataset.map(preprocess, num_parallel_calls=tf.data.AUTOTUNE)
    dataset = dataset.batch(batch_size)
    dataset = dataset.prefetch(tf.data.AUTOTUNE)
    return dataset

# --- Fonctions loss et metrics ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)

def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)

def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)

# --- Modèle DeepLabV3+ ---
def deeplabv3plus_resnet101(image_size=(256, 256), num_classes=8, weights_path=None, backbone_weights_path=None):
    inputs = Input(shape=(*image_size, 3))
    base_model = applications.ResNet101(input_tensor=inputs, include_top=False, weights=None)
    if backbone_weights_path:
        base_model.load_weights(backbone_weights_path)

    low_level_feat = base_model.get_layer('conv2_block3_out').output
    low_level_feat = Conv2D(48, 1, padding='same', activation='relu')(low_level_feat)

    x = base_model.output
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = UpSampling2D(size=(8, 8), interpolation='bilinear')(x)

    x = Concatenate()([x, low_level_feat])
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = Conv2D(256, 3, padding='same', activation='relu')(x)

    x = UpSampling2D(size=(4, 4), interpolation='bilinear')(x)
    outputs = Conv2D(num_classes, 1, padding='same', activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)
    if weights_path:
        model.load_weights(weights_path)
    return model

# --- Fonction d'entraînement ---
def train_deeplabv3plus_resnet101_local(
    image_size=(256,256),
    batch_size=2,
    epochs=5,
    sample_size=10000,
    data_dir="/path/to/data",
    model_weights_path=None,
    backbone_weights_path=None,
    model_name="deeplabv3plus_cityscapes_model_resnet101",
    experiment_name="deeplabv3plus_resnet101_local_exp"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit_aug/train"),
        os.path.join(data_dir, "masks_8classes_aug/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    with mlflow.start_run():
        print("🔌 Construction du modèle...")
        model = deeplabv3plus_resnet101(
            image_size=image_size,
            num_classes=8,
            weights_path=model_weights_path,
            backbone_weights_path=backbone_weights_path
        )
        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history_all = []
        for epoch in range(epochs):
            print(f"\n🔁 Epoch {epoch+1}/{epochs} : échantillon aléatoire de {sample_size} images")
            sampled_train_ds = create_sampled_dataset(train_images, train_masks, sample_size, batch_size, image_size=image_size)
            history = model.fit(sampled_train_ds, validation_data=(val_images, val_masks), epochs=1)
            history_all.append(history)

        print("📀 Sauvegarde du modèle...")
        model_path = "deeplabv3plus_resnet101_local.keras"
        model.save(model_path)
        mlflow.log_artifact(model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        print("📦 Enregistrement MLflow...")
        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name="deeplabv3plus_resnet101_local",
            input_example=input_example,
            signature=signature
        )

    return model, history_all
```

```python
import os
import numpy as np
from PIL import Image
from glob import glob
from concurrent.futures import ThreadPoolExecutor
from functools import partial
import tensorflow as tf
from tensorflow.keras import layers, models, applications
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, UpSampling2D, Concatenate

# --- Chargement des images (en mémoire, toutes les images) ---
def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)

def load_images_and_masks(image_dir, mask_dir, image_size=(256, 256), num_workers=40):
    image_paths = sorted(glob(os.path.join(image_dir, '*/*.png')))
    mask_paths = sorted(glob(os.path.join(mask_dir, '*/*_gtFine_8class.png')))

    print(f"📂 {len(image_paths)} images trouvées dans {image_dir}")
    print(f"📂 {len(mask_paths)} masques trouvés dans {mask_dir}")

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), image_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.array(masks, dtype=np.uint8)
    masks = np.expand_dims(masks, axis=-1)

    return images, masks

# --- DataGenerator avec sous-échantillonnage aléatoire à chaque epoch ---
class RandomSubsetGenerator(tf.keras.utils.Sequence):
    def __init__(self, images, masks, batch_size=8, sample_size=10000):
        self.images = images
        self.masks = masks
        self.batch_size = batch_size
        self.sample_size = sample_size
        self.indices = np.arange(len(images))
        self.on_epoch_end()

    def __len__(self):
        return self.sample_size // self.batch_size

    def __getitem__(self, index):
        batch_ids = self.subset_indices[index * self.batch_size:(index + 1) * self.batch_size]
        return self.images[batch_ids], self.masks[batch_ids]

    def on_epoch_end(self):
        self.subset_indices = np.random.choice(self.indices, size=self.sample_size, replace=False)

# --- Fonctions personnalisées ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    return tf.reduce_sum(loss * tf.cast(mask, loss.dtype)) / (tf.reduce_sum(tf.cast(mask, loss.dtype)) + 1e-8)

def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    iou_list = []
    for i in range(8):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou_list.append(tf.math.divide_no_nan(intersection, union))
    return tf.reduce_mean(iou_list)

def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)
    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])
    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)
    dice_list = []
    for i in range(8):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)

# --- Modèle DeepLabV3+ avec ResNet101 ---
def deeplabv3plus_resnet101(image_size=(256, 256), num_classes=8, weights_path=None, backbone_weights_path=None):
    inputs = Input(shape=(*image_size, 3))
    base_model = applications.ResNet101(include_top=False, weights=None, input_tensor=inputs)
    if backbone_weights_path:
        base_model.load_weights(backbone_weights_path)

    low_level_feat = base_model.get_layer('conv2_block3_out').output
    low_level_feat = Conv2D(48, 1, padding='same', activation='relu')(low_level_feat)

    x = base_model.output
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = UpSampling2D(size=(8, 8), interpolation='bilinear')(x)
    x = Concatenate()([x, low_level_feat])
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = Conv2D(256, 3, padding='same', activation='relu')(x)
    x = UpSampling2D(size=(4, 4), interpolation='bilinear')(x)
    outputs = Conv2D(num_classes, 1, padding='same', activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)
    if weights_path:
        model.load_weights(weights_path)
    return model

# --- Entraînement avec MLflow ---
def train_deeplabv3plus_resnet101_local(
    image_size=(256, 256),
    batch_size=8,
    epochs=10,
    sample_size=10000,
    data_dir="/path/to/data",
    model_weights_path=None,
    backbone_weights_path=None,
     model_name="exp_deeplabv3plus_resnet101_local_256x256_data_augmented_sample_10K_shuffle",
    experiment_name="deeplabv3plus_resnet101_exp"
):
    import mlflow
    import mlflow.tensorflow
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données...")
    train_images, train_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit_aug/train"),
        os.path.join(data_dir, "masks_8classes_aug/train"),
        image_size=image_size
    )
    val_images, val_masks = load_images_and_masks(
        os.path.join(data_dir, "leftImg8bit/val"),
        os.path.join(data_dir, "masks_8classes/val"),
        image_size=image_size
    )

    train_gen = RandomSubsetGenerator(train_images, train_masks, batch_size=batch_size, sample_size=sample_size)

    with mlflow.start_run():
        print("🧠 Construction du modèle...")
        model = deeplabv3plus_resnet101(
            image_size=image_size,
            weights_path=model_weights_path,
            backbone_weights_path=backbone_weights_path
        )

        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement...")
        history = model.fit(
            train_gen,
            validation_data=(val_images, val_masks),
            epochs=epochs
        )

        print("💾 Sauvegarde...")
        keras_model_path = "deeplabv3plus_resnet101_local.keras"
        model.save(keras_model_path)
        mlflow.log_artifact(keras_model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name="deeplabv3plus_resnet101_local",
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
model, history = train_deeplabv3plus_resnet101_local(
    image_size=(256, 256),
    batch_size=10,
    epochs=2,
    data_dir="/srv/picture_segmentation/projet8_openclassrooms/notebooks/data",
    model_weights_path=None,
    backbone_weights_path="/srv/picture_segmentation/projet8_openclassrooms/notebooks/resnet101_weights_tf_dim_ordering_tf_kernels_notop.weights.h5",
    experiment_name="exp_deeplabv3plus_resnet101_local_data_augmented_sample_10K_shuffle",
    model_name="exp_deeplabv3plus_resnet101_local_256x256_data_augmented_sample_10K_shuffle",
    sample_size=10000,
)
```

# Entrainement 1024 x 2048 sur sous ensembles équilibrés

```python
import os
import numpy as np
from PIL import Image
from glob import glob
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from tensorflow.keras.models import Model
from tensorflow.keras.layers import Input, Conv2D, MaxPooling2D, UpSampling2D, concatenate, BatchNormalization
from tensorflow.keras.optimizers import Adam
from concurrent.futures import ThreadPoolExecutor
from functools import partial

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"
tf.get_logger().setLevel("ERROR")


def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize(image_size[::-1])
    mask = Image.open(mask_path).convert("L").resize(image_size[::-1], resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)


def load_subset_images_and_masks(subset_path, image_size, num_workers):
    img_dir = os.path.join(subset_path, "images")
    mask_dir = os.path.join(subset_path, "masks")

    img_paths = sorted(glob(os.path.join(img_dir, "*/*.png")))
    mask_paths = sorted(glob(os.path.join(mask_dir, "*/*_gtFine_8class.png")))

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), img_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.expand_dims(np.array(masks, dtype=np.uint8), axis=-1)
    return images, masks


def load_all_subsets(subsets_root, image_size, num_workers):
    subset_dirs = sorted([os.path.join(subsets_root, d) for d in os.listdir(subsets_root) if os.path.isdir(os.path.join(subsets_root, d))])
    all_images, all_masks = [], []

    for subset in subset_dirs:
        print(f"📂 Chargement {subset} ...")
        imgs, msks = load_subset_images_and_masks(subset, image_size, num_workers)
        all_images.append(imgs)
        all_masks.append(msks)

    return np.concatenate(all_images, axis=0), np.concatenate(all_masks, axis=0)


def load_val_images_and_masks(val_img_dir, val_mask_dir, image_size, num_workers):
    img_paths = sorted(glob(os.path.join(val_img_dir, "*/*.png")))
    mask_paths = sorted(glob(os.path.join(val_mask_dir, "*/*_gtFine_8class.png")))

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), img_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.expand_dims(np.array(masks, dtype=np.uint8), axis=-1)
    return images, masks


# --- U-Net ---
def double_conv(x, filters):
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    x = Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = BatchNormalization()(x)
    return x


def build_unet(input_shape=(1024, 2048, 3), num_classes=8, depth=4, base_filters=64):
    inputs = Input(shape=input_shape)
    encoders = []
    x = inputs

    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = MaxPooling2D()(x)

    x = double_conv(x, base_filters * (2 ** depth))

    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = UpSampling2D()(x)
        x = concatenate([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = Conv2D(num_classes, 1, activation='softmax')(x)
    return Model(inputs, outputs)


# --- Loss & Metrics ---
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    return tf.reduce_sum(loss * tf.cast(mask, loss.dtype)) / (tf.reduce_sum(tf.cast(mask, loss.dtype)) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    ious = []
    for i in range(8):
        intersect = tf.reduce_sum(tf.cast((y_true == i) & (y_pred == i), tf.float32))
        union = tf.reduce_sum(tf.cast((y_true == i) | (y_pred == i), tf.float32))
        ious.append(tf.math.divide_no_nan(intersect, union))
    return tf.reduce_mean(ious)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.boolean_mask(tf.reshape(y_true, [-1]), tf.reshape(mask, [-1]))
    y_pred_flat = tf.boolean_mask(tf.reshape(y_pred, [-1]), tf.reshape(mask, [-1]))

    dice_scores = []
    for i in range(8):
        y_true_c = tf.cast(y_true_flat == i, tf.float32)
        y_pred_c = tf.cast(y_pred_flat == i, tf.float32)
        inter = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * inter) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_scores.append(dice)
    return tf.reduce_mean(dice_scores)


# --- Entraînement principal ---
def train_unet_from_subsets(
    subsets_root,
    val_images_dir,
    val_masks_dir,
    image_size=(1024, 2048),
    batch_size=2,
    epochs=10,
    depth=4,
    base_filters=64,
    num_workers=8,
    experiment_name="unet-segmentation",
    model_name="unet_model"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    print("📦 Chargement des données d'entraînement...")
    train_images, train_masks = load_all_subsets(subsets_root, image_size, num_workers)
    print(f"✅ {train_images.shape[0]} images d'entraînement chargées.")

    print("📦 Chargement des données de validation...")
    val_images, val_masks = load_val_images_and_masks(val_images_dir, val_masks_dir, image_size, num_workers)
    print(f"✅ {val_images.shape[0]} images de validation chargées.")

    with mlflow.start_run():
        model = build_unet(input_shape=(*image_size, 3), num_classes=8, depth=depth, base_filters=base_filters)
        model.compile(
            optimizer=Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        print("🚀 Entraînement en cours...")
        history = model.fit(
            train_images, train_masks,
            validation_data=(val_images, val_masks),
            epochs=epochs,
            batch_size=batch_size
        )

        model_path = f"{model_name}.keras"
        model.save(model_path)
        mlflow.log_artifact(model_path)

        input_example = train_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model, history
```

```python
import os
import numpy as np
import tensorflow as tf
import mlflow
import mlflow.tensorflow
from glob import glob
from PIL import Image
from concurrent.futures import ThreadPoolExecutor
from functools import partial
import gc


# --------------------------
# UTILITAIRES DE CHARGEMENT
# --------------------------
def _load_and_process_pair(img_path, mask_path, image_size):
    img = Image.open(img_path).convert("RGB").resize(image_size[::-1])
    mask = Image.open(mask_path).convert("L").resize(image_size[::-1], resample=Image.NEAREST)
    return np.array(img) / 255.0, np.array(mask)


def load_subset_images_and_masks(subset_path, image_size, num_workers):
    img_dir = os.path.join(subset_path, "images")
    mask_dir = os.path.join(subset_path, "masks")

    img_paths = sorted(glob(os.path.join(img_dir, "*/*.png")))
    mask_paths = sorted(glob(os.path.join(mask_dir, "*/*_gtFine_8class.png")))

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), img_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.expand_dims(np.array(masks, dtype=np.uint8), axis=-1)
    return images, masks


def load_val_images_and_masks(val_img_dir, val_mask_dir, image_size, num_workers):
    img_paths = sorted(glob(os.path.join(val_img_dir, "*/*.png")))
    mask_paths = sorted(glob(os.path.join(val_mask_dir, "*/*_gtFine_8class.png")))

    with ThreadPoolExecutor(max_workers=num_workers) as executor:
        results = list(executor.map(partial(_load_and_process_pair, image_size=image_size), img_paths, mask_paths))

    images, masks = zip(*results)
    images = np.array(images, dtype=np.float32)
    masks = np.expand_dims(np.array(masks, dtype=np.uint8), axis=-1)
    return images, masks


# --------------------------
# U-Net
# --------------------------
def double_conv(x, filters):
    x = tf.keras.layers.Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    x = tf.keras.layers.Conv2D(filters, 3, activation='relu', padding='same')(x)
    x = tf.keras.layers.BatchNormalization()(x)
    return x


def build_unet(input_shape, num_classes=8, depth=4, base_filters=64):
    inputs = tf.keras.Input(shape=input_shape)
    encoders = []
    x = inputs

    for i in range(depth):
        filters = base_filters * (2 ** i)
        x = double_conv(x, filters)
        encoders.append(x)
        x = tf.keras.layers.MaxPooling2D()(x)

    x = double_conv(x, base_filters * (2 ** depth))

    for i in reversed(range(depth)):
        filters = base_filters * (2 ** i)
        x = tf.keras.layers.UpSampling2D()(x)
        x = tf.keras.layers.Concatenate()([x, encoders[i]])
        x = double_conv(x, filters)

    outputs = tf.keras.layers.Conv2D(num_classes, 1, activation='softmax')(x)
    return tf.keras.Model(inputs, outputs)


# --------------------------
# Loss & Metrics
# --------------------------
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    return tf.reduce_sum(loss * tf.cast(mask, loss.dtype)) / (tf.reduce_sum(tf.cast(mask, loss.dtype)) + 1e-8)


def masked_mean_iou(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    ious = []
    for i in range(8):
        intersect = tf.reduce_sum(tf.cast((y_true == i) & (y_pred == i), tf.float32))
        union = tf.reduce_sum(tf.cast((y_true == i) | (y_pred == i), tf.float32))
        ious.append(tf.math.divide_no_nan(intersect, union))
    return tf.reduce_mean(ious)


def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.boolean_mask(tf.reshape(y_true, [-1]), tf.reshape(mask, [-1]))
    y_pred_flat = tf.boolean_mask(tf.reshape(y_pred, [-1]), tf.reshape(mask, [-1]))

    dice_scores = []
    for i in range(8):
        y_true_c = tf.cast(y_true_flat == i, tf.float32)
        y_pred_c = tf.cast(y_pred_flat == i, tf.float32)
        inter = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * inter) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_scores.append(dice)
    return tf.reduce_mean(dice_scores)


# --------------------------
# TRAINING PAR SUBSET
# --------------------------
def train_unet_from_subsets_sequential(
    subsets_root,
    val_images_dir,
    val_masks_dir,
    image_size=(1024, 2048),
    batch_size=2,
    epochs=10,
    depth=4,
    base_filters=64,
    num_workers=8,
    experiment_name="unet-subset-cycling",
    model_name="unet_sequential"
):
    from mlflow.models.signature import infer_signature

    mlflow.set_tracking_uri("http://127.0.0.1:8080")
    mlflow.set_experiment(experiment_name)
    mlflow.tensorflow.autolog()

    subset_dirs = sorted([os.path.join(subsets_root, d)
                          for d in os.listdir(subsets_root)
                          if os.path.isdir(os.path.join(subsets_root, d))])
    n_subsets = len(subset_dirs)

    print(f"🔍 {n_subsets} sous-ensembles détectés.")

    print("📦 Chargement du jeu de validation...")
    val_images, val_masks = load_val_images_and_masks(val_images_dir, val_masks_dir, image_size, num_workers)

    with mlflow.start_run():
        model = build_unet(input_shape=(*image_size, 3), num_classes=8, depth=depth, base_filters=base_filters)
        model.compile(
            optimizer=tf.keras.optimizers.Adam(1e-3),
            loss=sparse_categorical_crossentropy_ignore_255,
            metrics=["accuracy", masked_mean_iou, masked_dice_coef]
        )

        for epoch in range(epochs):
            subset_idx = epoch % n_subsets
            subset_path = subset_dirs[subset_idx]
            print(f"\n🌀 Epoch {epoch+1}/{epochs} → Chargement du subset {subset_idx + 1}: {subset_path}")
            train_images, train_masks = load_subset_images_and_masks(subset_path, image_size, num_workers)

            model.fit(train_images, train_masks,
                      validation_data=(val_images, val_masks),
                      epochs=1,
                      batch_size=batch_size)

            # Libération mémoire
            del train_images, train_masks
            gc.collect()
            tf.keras.backend.clear_session()

        model_path = f"{model_name}.keras"
        model.save(model_path)
        mlflow.log_artifact(model_path)

        # Log signature
        input_example = val_images[:1]
        pred_example = model.predict(input_example)
        signature = infer_signature(input_example, pred_example)

        mlflow.tensorflow.log_model(
            model=model,
            artifact_path="model",
            registered_model_name=model_name,
            input_example=input_example,
            signature=signature
        )

    return model
```

```python
import tensorflow as tf
from keras.saving import register_keras_serializable

@register_keras_serializable()
def sparse_categorical_crossentropy_ignore_255(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true_clean = tf.where(mask, y_true, tf.zeros_like(y_true))
    loss = tf.keras.losses.sparse_categorical_crossentropy(y_true_clean, y_pred)
    mask = tf.cast(mask, loss.dtype)
    loss = loss * mask
    return tf.reduce_sum(loss) / (tf.reduce_sum(mask) + 1e-8)

@register_keras_serializable()
def masked_mean_iou(y_true, y_pred):
    num_classes = 8
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    iou_list = []
    for i in range(num_classes):
        y_true_i = tf.equal(y_true, i)
        y_pred_i = tf.equal(y_pred, i)
        intersection = tf.reduce_sum(tf.cast(y_true_i & y_pred_i, tf.float32))
        union = tf.reduce_sum(tf.cast(y_true_i | y_pred_i, tf.float32))
        iou = tf.math.divide_no_nan(intersection, union)
        iou_list.append(iou)
    return tf.reduce_mean(iou_list)

@register_keras_serializable()
def masked_dice_coef(y_true, y_pred):
    y_true = tf.squeeze(y_true, axis=-1)
    mask = tf.not_equal(y_true, 255)
    y_true = tf.where(mask, y_true, 0)
    y_pred = tf.argmax(y_pred, axis=-1, output_type=tf.int32)

    y_true_flat = tf.reshape(y_true, [-1])
    y_pred_flat = tf.reshape(y_pred, [-1])
    mask_flat = tf.reshape(mask, [-1])

    y_true_flat = tf.boolean_mask(y_true_flat, mask_flat)
    y_pred_flat = tf.boolean_mask(y_pred_flat, mask_flat)

    dice_list = []
    num_classes = 8
    for i in range(num_classes):
        y_true_c = tf.cast(tf.equal(y_true_flat, i), tf.float32)
        y_pred_c = tf.cast(tf.equal(y_pred_flat, i), tf.float32)
        intersection = tf.reduce_sum(y_true_c * y_pred_c)
        dice = (2. * intersection) / (tf.reduce_sum(y_true_c) + tf.reduce_sum(y_pred_c) + 1e-8)
        dice_list.append(dice)
    return tf.reduce_mean(dice_list)



from tensorflow.keras.models import load_model

model_path = "downloaded_model/data/model.keras"

model = load_model(
    model_path,
    custom_objects={
        "sparse_categorical_crossentropy_ignore_255": sparse_categorical_crossentropy_ignore_255,
        "masked_mean_iou": masked_mean_iou,
        "masked_dice_coef": masked_dice_coef
    }
)

print("✅ Modèle chargé avec succès")
```

```python
import numpy as np
from PIL import Image
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
import tensorflow as tf
import os

def overlay_mask_on_image_8class(image_pil, mask_np, alpha=0.6):
    """
    Version modifiée pour accepter PIL.Image et np.array en entrée,
    et afficher l'overlay.
    """

    # Palette (fluorescente) pour 8 classes : 0 = void
    class_colors = {
        0: ((0, 0, 0), "void"),
        1: ((255, 0, 0), "flat"),
        2: ((255, 128, 0), "construction"),
        3: ((255, 255, 0), "object"),
        4: ((0, 255, 0), "nature"),
        5: ((0, 255, 255), "sky"),
        6: ((0, 0, 255), "human"),
        7: ((255, 0, 255), "vehicle"),
    }

    # Construction du masque colorisé
    color_mask = np.zeros((*mask_np.shape, 3), dtype=np.uint8)
    legend_handles = []

    for class_id, (color, label) in class_colors.items():
        region = mask_np == class_id
        if np.any(region):
            color_mask[region] = color
            legend_handles.append(
                mpatches.Patch(color=np.array(color)/255.0, label=f"{class_id}: {label}")
            )

    color_mask_img = Image.fromarray(color_mask)

    # Resize masque colorisé au format de l'image d'origine
    if color_mask_img.size != image_pil.size:
        color_mask_img = color_mask_img.resize(image_pil.size, resample=Image.NEAREST)

    # Fusion avec alpha
    blended = Image.blend(image_pil.convert("RGBA"), color_mask_img.convert("RGBA"), alpha)

    # Affichage
    plt.figure(figsize=(10, 10))
    plt.imshow(blended)
    plt.title("Image avec masque 8 classes superposé")
    plt.axis("off")
    if legend_handles:
        plt.legend(
            handles=legend_handles,
            bbox_to_anchor=(1.05, 1),
            loc='upper left',
            borderaxespad=0.,
            fontsize='small'
        )
    plt.tight_layout()
    plt.show()


def predict_and_save_mask(
    model,
    image_path,
    output_mask_dir,
    input_size=(256, 512),
    save_mask_as_png=True
):
    """
    Prédit le masque pour une image originale, sauvegarde le masque redimensionné,
    puis affiche l'overlay.

    Args:
        model: modèle keras chargé.
        image_path (str): chemin vers l'image originale.
        output_mask_dir (str): dossier où sauvegarder le masque prédit.
        input_size (tuple): taille (hauteur, largeur) pour le modèle.
        save_mask_as_png (bool): si True, sauvegarde le masque en PNG (uint8).
    """

    os.makedirs(output_mask_dir, exist_ok=True)

    # Chargement image originale
    image_orig = Image.open(image_path).convert("RGB")
    orig_size = image_orig.size  # (largeur, hauteur)

    # Préparation image pour le modèle : resize + normalisation [0,1]
    image_resized = image_orig.resize((input_size[1], input_size[0]), Image.BILINEAR)
    input_array = np.array(image_resized) / 255.0
    input_array = np.expand_dims(input_array, axis=0)  # batch dimension

    # Prédiction masque
    pred = model.predict(input_array)[0]  # (256,512,num_classes)
    pred_mask = np.argmax(pred, axis=-1).astype(np.uint8)  # (256,512)

    # Remise à la taille originale
    pred_mask_pil = Image.fromarray(pred_mask, mode="L")
    pred_mask_orig_size = pred_mask_pil.resize(orig_size, resample=Image.NEAREST)
    pred_mask_orig_np = np.array(pred_mask_orig_size, dtype=np.uint8)

    # Sauvegarde masque
    base_name = os.path.basename(image_path)
    name_wo_ext = os.path.splitext(base_name)[0]
    mask_save_path = os.path.join(output_mask_dir, f"{name_wo_ext}_mask.png")
    if save_mask_as_png:
        Image.fromarray(pred_mask_orig_np).save(mask_save_path)
    else:
        np.save(os.path.join(output_mask_dir, f"{name_wo_ext}_mask.npy"), pred_mask_orig_np)

    print(f"Masque sauvegardé sous {mask_save_path}")

    # Affichage overlay
    overlay_mask_on_image_8class(image_orig, pred_mask_orig_np, alpha=0.6)


# --- Exemple d'utilisation ---

# model = load_model("ton_modele.keras", custom_objects={...})

# predict_and_save_mask(
#     model,
#     "chemin/vers/image_1024x2048.png",
#     "chemin/vers/dossier_masques_predits",
#     input_size=(256, 512)
# )
```

```python
import os
import numpy as np
from PIL import Image, ImageDraw, ImageFont
import matplotlib.pyplot as plt

def process_and_save_masked_overlay_with_legend(img_path, mask_path, output_dir, image_size=(256, 512), alpha=0.5):
    """
    Redimensionne l'image et le masque, applique le masque coloré,
    ajoute une légende à l'image, sauvegarde et affiche l'overlay.
    """
    os.makedirs(output_dir, exist_ok=True)

    # Chargement et redimensionnement
    img = Image.open(img_path).convert("RGB").resize((image_size[1], image_size[0]))
    mask = Image.open(mask_path).convert("L").resize((image_size[1], image_size[0]), resample=Image.NEAREST)

    img_np = np.array(img) / 255.0
    mask_np = np.array(mask)

    # Colormap + labels
    colormap = {
        0: ((0, 0, 0), "void"),
        1: ((255, 0, 0), "flat"),
        2: ((255, 128, 0), "construction"),
        3: ((255, 255, 0), "object"),
        4: ((0, 255, 0), "nature"),
        5: ((0, 255, 255), "sky"),
        6: ((0, 0, 255), "human"),
        7: ((255, 0, 255), "vehicle"),
    }

    # Masque coloré
    color_mask = np.zeros((*mask_np.shape, 3), dtype=np.uint8)
    for class_id, (color, _) in colormap.items():
        color_mask[mask_np == class_id] = color
    color_mask = color_mask.astype(np.float32) / 255.0

    # Overlay image
    overlay = (1 - alpha) * img_np + alpha * color_mask
    overlay_img = Image.fromarray((overlay * 255).astype(np.uint8))

    # Ajouter légende sur la droite
    legend_width = 150
    final_width = overlay_img.width + legend_width
    final_height = overlay_img.height

    legend_img = Image.new("RGB", (final_width, final_height), (255, 255, 255))
    legend_img.paste(overlay_img, (0, 0))

    draw = ImageDraw.Draw(legend_img)

    try:
        font = ImageFont.truetype("arial.ttf", 14)
    except:
        font = ImageFont.load_default()

    y_offset = 10
    box_size = 20
    spacing = 8

    for class_id, (rgb, label) in colormap.items():
        y = y_offset + class_id * (box_size + spacing)
        draw.rectangle([overlay_img.width + 10, y, overlay_img.width + 10 + box_size, y + box_size], fill=rgb)
        draw.text((overlay_img.width + 10 + box_size + 5, y), label, fill=(0, 0, 0), font=font)

    # Sauvegarde
    base_name = os.path.splitext(os.path.basename(img_path))[0]
    output_path = os.path.join(output_dir, f"{base_name}_overlay_legend.png")
    legend_img.save(output_path)
    print(f"✅ Sauvegardé avec légende : {output_path}")

    # Affichage
    plt.figure(figsize=(10, 5))
    plt.imshow(legend_img)
    plt.axis("off")
    plt.title("Overlay avec légende")
    plt.show()
```

```python
process_and_save_masked_overlay_with_legend(
    img_path="data/leftImg8bit/val/frankfurt/frankfurt_000001_083199_leftImg8bit.png",
    mask_path="data/masks_8_true/val/frankfurt/frankfurt_000001_083199_gtFine_8class.png",
    output_dir="outputs/overlays",  # dossier cible
    image_size=(256, 512)
)
```
